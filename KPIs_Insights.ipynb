```python
import pandas as pd

# Load the cleaned dataset
df = pd.read_csv('cleaned_titanic.csv')

# --- 1. Calculate Overall KPIs ---
total_passengers = len(df)
total_survived = df['Survived'].sum()
overall_survival_rate = round(df['Survived'].mean() * 100, 2)
avg_fare = round(df['Fare'].mean(), 2)
avg_age = round(df['Age'].mean(), 1)
```

```python
print("=== OVERALL KPIs ===")
print(f"Total Passengers: {total_passengers}")
print(f"Total Survived: {total_survived}")
print(f"Overall Survival Rate: {overall_survival_rate}%")
print(f"Average Fare: ${avg_fare}")
print(f"Average Age: {avg_age} years")
```

```python
# --- 2. Survival Rate by Categorical Variables ---
# Survival by Sex
survival_by_sex = df.groupby('Sex')['Survived'].agg(['count', 'mean'])
survival_by_sex['mean'] = (survival_by_sex['mean'] * 100).round(2)
```

```python
# Survival by Class
survival_by_class = df.groupby('Pclass')['Survived'].agg(['count', 'mean'])
survival_by_class['mean'] = (survival_by_class['mean'] * 100).round(2)
```

```python
# Survival by Solitary Status (Alone vs Family)
survival_by_alone = df.groupby('IsAlone')['Survived'].agg(['count', 'mean'])
survival_by_alone['mean'] = (survival_by_alone['mean'] * 100).round(2)
```

```python
# Survival by Cabin Availability
survival_by_cabin = df.groupby('Has_Cabin')['Survived'].agg(['count', 'mean'])
survival_by_cabin['mean'] = (survival_by_cabin['mean'] * 100).round(2)
```

```python
print("\n=== SURVIVAL BY SEX (%) ===")
print(survival_by_sex)

print("\n=== SURVIVAL BY PCLASS (%) ===")
print(survival_by_class)

print("\n=== SURVIVAL BY IS_ALONE (%) ===")
print(survival_by_alone)
```
